## Point config at your trained models

If you trained in Colab Phase 1/2, set paths (adjust as needed):


````python
from promptforge.local_paths import save_user_config, load_user_config
from pathlib import Path

quality = Path("/content/promptforge-quality-model")
optimizer = Path("/content/promptforge-optimizer-model")

# Fall back to repo outputs for local runs
if not quality.exists():
    quality = REPO / "outputs" / "promptforge-quality-model"
if not optimizer.exists():
    optimizer = REPO / "outputs" / "promptforge-optimizer-model"

cfg = load_user_config()
if quality.exists():
    cfg["quality_model_path"] = str(quality)
if optimizer.exists():
    cfg["optimizer_model_path"] = str(optimizer)
cfg["prefer_gpu"] = True
path = save_user_config(cfg)
print("Wrote", path)
print(cfg)

!promptforge doctor
````

````python
!promptforge analyze "Build me a website"
print("---")
!promptforge run "Make an app." --json
````

````python
from promptforge import PromptForge
import json

pf = PromptForge.from_config()
print("quality:", pf.quality_model_path)
print("optimizer:", pf.optimizer_model_path)

print(json.dumps(pf.analyze("Build me a website"), indent=2))
if pf.optimizer is not None and pf.scorer is not None:
    print(json.dumps(pf.run("Build me a website"), indent=2))
````
